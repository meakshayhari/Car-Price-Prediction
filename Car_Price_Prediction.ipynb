```python
def compute_cost(X, y, w, b):
    m = len(y)
    predictions = X @ w + b
    errors = predictions - y
    cost = np.sum(errors ** 2) / (2 * m)
    return cost
```

```python
def compute_gradient(X, y, w, b):
    m = len(y)
    errors = (X @ w + b) - y
    dj_dw = (X.T @ errors) / m
    dj_db = np.sum(errors) / m
    return dj_dw, dj_db
```

```python

def gradient_descent(X, y, w, b, alpha, num_iterations):
    J_history = []

    for i in range(num_iterations):
        dj_dw, dj_db = compute_gradient(X, y, w, b)

        w = w - alpha * dj_dw
        b = b - alpha * dj_db

        J_history.append(compute_cost(X, y, w, b))

    return w, b, J_history
```

```python

w, b, J_history = gradient_descent(
    X_train, y_train,
    np.zeros(X_train.shape[1]), 0,
    alpha=0.01,
    num_iterations=1000
)

predictions = X_test @ w + b

test_errors = predictions - y_test
mae = np.mean(np.abs(test_errors))
rmse = np.sqrt(np.mean(test_errors ** 2))

print("Initial cost:", J_history[0])
print("Final cost:", J_history[-1])
print("Test MAE:", mae)
print("Test RMSE:", rmse)
```
